#### Assignment of Plotting
Refer to the tips.csv dataset. The dataset contains total 244 tips(data about customers
served by a restaurant) across 7 different columns.
 totbill - Total bill (cost of the meal), including tax, in US dollars
 tip - Tip (gratuity) in US dollars
 sex - Sex of person paying for the meal (0=male, 1=female)
 smoker - Smoker in party? (0=No, 1=Yes)
 day - Sun, Sat, Thu, Fri
 time - Night and Day
 size - Size of the party
1. Plot a day wise boxplot of totbill column.

2. Plot a pariplot by setting hue parameter t

3. Draw a scatter plot of tip and totalbill and also allot the points according to the sex
4. Plot a hitsogram of totbill for 50 bins
5. Plot a heatmap to represent entire dataset's correlation

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("tips.csv")

print(df.head())
print(df.columns)
```

```python
#1. Day-wise Boxplot of Total Bill
sns.boxplot(x="day", y="total_bill", data=df)

plt.title("Day-wise Boxplot of Total Bill")
plt.xlabel("Day")
plt.ylabel("Total Bill")

plt.show()
```

```python
#2. Pairplot by setting Hue

sns.pairplot(df, hue="sex")

plt.show()
```

```python
##3. Scatter Plot of Tip and Total Bill according to Sex
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=df)

plt.title("Tip vs Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

```python
##4. Histogram of Total Bill for 50 Bins
plt.hist(df["total_bill"], bins=50)

plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
#5. Heatmap of Entire Dataset Correlation
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
