#### 2. Write code to replace the odd numbers by -1 in the array created by using
random.randint( 1 , 50,20)

```python
import numpy as np
```

```python
a=np.random.randint( 1 , 50,20)
```

```python
print("Original Array:")
print(a)
```

```python
a[a % 2 != 0] = -1

print("After replacing odd numbers:")
print(a)
```
