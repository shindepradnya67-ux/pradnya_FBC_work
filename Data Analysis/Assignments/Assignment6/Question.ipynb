```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Cars93.csv")

print(df.head())
```

```python
#1. Bar Chart – Average Price by Manufacturer
#Code:
avg_price = df.groupby("Manufacturer")["Price"].mean()

plt.figure(figsize=(12,6))

plt.bar(avg_price.index, avg_price.values)

plt.xlabel("Manufacturer")
plt.ylabel("Average Price")
plt.title("Average Car Price by Manufacturer")

plt.xticks(rotation=90)

plt.show()
```

```python
#2. Line Chart – Horsepower for Different Types of Cars

#इथे Type म्हणजे car type वापरला आहे.

type_hp = df.groupby("Type")["Horsepower"].mean()

plt.figure(figsize=(10,6))

plt.plot(type_hp.index, type_hp.values, marker="o")

plt.xlabel("Car Type")
plt.ylabel("Average Horsepower")
plt.title("Average Horsepower by Car Type")

plt.legend(["Horsepower"])

plt.show()
```

```python
#3. Pie Chart – Proportion of Different Car Types
type_count = df["Type"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    type_count.values,
    labels=type_count.index,
    autopct="%1.1f%%"
)

plt.title("Proportion of Different Car Types")

plt.show()
```

```python
##4. Scatter Plot – Horsepower vs MPG
plt.figure(figsize=(8,6))

plt.scatter(df["Horsepower"], df["MPG.city"])

plt.xlabel("Horsepower")
plt.ylabel("MPG City")
plt.title("Relationship Between Horsepower and MPG")

plt.show()
```

```python
#5. Line Chart – Horsepower for Different Types of Cars

#Question 5 हे Question 2 सारखेच आहे. Assignment मध्ये repeat झाले आहे.

#ड्या वेगळ्या पद्धतीने:

types = df["Type"].unique()

for car_type in types:
    data = df[df["Type"] == car_type]
    plt.plot(
        data["Manufacturer"],
        data["Horsepower"],
        marker="o",
        label=car_type
    )

plt.xlabel("Manufacturer")
plt.ylabel("Horsepower")
plt.title("Horsepower Trend for Different Car Types")

plt.legend()
plt.xticks(rotation=90)

plt.show()
```
