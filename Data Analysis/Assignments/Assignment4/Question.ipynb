```python
##1. Create a DataFrame:
#o Load the above dataset into a Pandas DataFrame.
```

```python
import pandas as pd
```

```python
data={
    "Order ID":[1001,1002,1003,1004,1005,1006,1007,1008],
    "Customer":["Alice","Bob","Charlie","Alice","David","Charlie","Bob","Alice"],
    "Product":["Laptop","Phone","chair","Table","Laptop","Phone","Chair","Phone"],
    "Category":["Electronics","Electronics","Furniture","Furniture","Electronics","Electronics","Furniture","Electronics"],
    "Quantity":[2,1,4,1,1,2,2,1],
    "Price":[800,500,100,250,800,500,100,500],
    "Total Sales":[1600,500,400,250,800,1000,200,500]
}

df=pd.DataFrame(data)
df
```

```python
##2. Add a New Column:
#o Add a column Discount that gives a 10% discount on total
#sales.
```

```python
df["Discount"]=df["Total Sales"]*0.10
df
```

```python
##3. Add a New Row:
# o Add a new sales record: Order ID = 1009, Customer =
# "Eve", Product = "Tablet", Category = "Electronics",
# Quantity = 1, Price = 300, Total Sales = 300.
```

```python
new_row={
    "Order ID":1009,
    "Customer":"Eve",
    "Product":"Table",
    "Category":"Electronics",
    "Quantity":1,
    "Price":300,
    "Total Sales":300,
    "Discount":300*0.10
}
df.loc[len(df)]=new_row
df
```

```python
# ##4. Filter Data:
# o Retrieve only the sales records where the Category is
# "Electronics".
```

```python
electronics=df[df["Category"]=="Electronics"]
electronics
```

```python
# 4. Filter Data:
# o Retrieve only the sales records where the Category is
# "Electronics".
```

```python
electronics=df[df["Category"]=="Electronics"]
electronics
```

```python
# #5.5. Selection:
# o Select the Customer and Total Sales columns.
```

```python
selected=df[["Customer","Total Sales"]]
selected
```

```python
# ##6.6. Grouping:
# o Group the data by Customer and calculate the total sales per
# customer.
```

```python
customer_sales=df.groupby("Customer")["Total Sales"].sum()
customer_sales
```

```python
# ##7. Sorting:
# o Sort the DataFrame based on Total Sales in descending
# order.
```

```python
sorted_df=df.sort_values(by="Total Sales",
                         ascending=False)
sorted_df
```

```python
# ##8. Apply Function:
# o Apply a lambda function to create a new column Sales
# Category, where values are labeled as "High" if Total Sales >
# 1000, otherwise "Low".
```

```python
df["Sales Category"]=df["Total Sales"].apply(lambda x:"High" if x>1000 else "Low")
df
```

```python
# 9. Save DataFrame:
# o Save the final DataFrame as a CSV file named sales_data.csv.
```

```python
df.to_csv("sales_data.csv",index=False)
print("DataFrame saved successfully as sales_data.csv")
```
